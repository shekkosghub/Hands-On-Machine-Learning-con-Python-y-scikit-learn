```python
# ---------------------------------------------------------------
# CAPÍTULO 2/Pandas — Ejercicio 1
# ---------------------------------------------------------------

import seaborn as sns
titanic = sns.load_dataset("titanic")
# a) Shape
# print(f"Filas: {titanic.shape[0]} | Columnas: {titanic.shape[1]}")
# b) Columnas con valores faltantes
faltantes = titanic.isnull().sum()
print("\nColumnas con NaN:")
print(faltantes[faltantes > 0])
# c) Edad promedio
print(f"\nEdad promedio: {titanic['age'].mean():.2f} años")
# d) Porcentaje que sobrevivió
pct_surv = titanic["survived"].mean() * 100
print(f"Sobrevivientes: {pct_surv:.1f}%")
```

```python
# ---------------------------------------------------------------
# CAPÍTULO 2/Pandas — Ejercicio 2
# ---------------------------------------------------------------

import seaborn as sns
import numpy as np

titanic = sns.load_dataset("titanic")
# a) Eliminar los 2 NaN de embarked
titanic = titanic.dropna(subset=["embarked"])
# b) Imputar age con la mediana
mediana_age = titanic["age"].median()
titanic["age"] = titanic["age"].fillna(mediana_age)

# c) Nueva columna familia
titanic["familia"] = titanic["sibsp"] + titanic["parch"]
# d) Codificar sex como numérico
titanic["sex_num"] = titanic["sex"].map({"male": 0, "female": 1})
# Seleccionar features finales
features = ["pclass", "sex_num", "age", "fare", "familia"]
X = titanic[features]
# Verificar que no queden NaN
assert X.isnull().sum().sum() == 0, "Aún hay NaN"
X_array = X.to_numpy()
print(f"Shape final: {X_array.shape}")
print(f"Tipo:        {X_array.dtype}")
```
