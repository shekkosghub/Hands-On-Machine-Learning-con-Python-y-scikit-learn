```python
# ---------------------------------------------------------------
# CAPÍTULO 2/Matplotlib y Seaborn — Ejercicio 1
# ---------------------------------------------------------------

import seaborn as sns
import matplotlib.pyplot as plt
titanic = sns.load_dataset("titanic")
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(titanic["fare"], bins=40, kde=True, ax=ax)
ax.set_title("Distribución de la tarifa (fare)")
ax.set_xlabel("Tarifa pagada (£)")
ax.set_ylabel("Frecuencia")
plt.tight_layout()
fig.savefig("distribucion_fare.png", dpi=150)
plt.show()
```

```python
# ---------------------------------------------------------------
# CAPÍTULO 2/Matplotlib y Seaborn — Ejercicio 2
# ---------------------------------------------------------------

import seaborn as sns
import matplotlib.pyplot as plt
titanic = sns.load_dataset("titanic").dropna(subset=["embarked"])
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# Panel izquierdo: boxplot fare por pclass
sns.boxplot(data=titanic, x="pclass", y="fare", ax=axes[0])
axes[0].set_title("Tarifa por clase")

# Panel derecho: countplot embarked con hue=survived
sns.countplot( data=titanic, x="embarked", hue="survived", ax=axes[1] )
axes[1].set_title("Embarcación y supervivencia")
fig.suptitle("Análisis exploratorio — Titanic", fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------------
# CAPÍTULO 2/Matplotlib y Seaborn — Ejercicio 3
# ---------------------------------------------------------------

import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

def eda_rapido(df, features, target):
    """
    Genera visualizaciones de análisis exploratorio.
    Parámetros
    ----------
    df       : pd.DataFrame
    features : list — columnas numéricas a analizar
    target   : str  — columna objetivo
    """
    carpeta = Path('figuras')
    carpeta.mkdir(exist_ok=True)

    # 1. Histograma por feature
    for feat in features:
        fig, ax = plt.subplots(figsize=(6, 4))
        sns.histplot(df[feat].dropna(), kde=True, ax=ax)
        ax.set_title(f"Distribución: {feat}")
        ax.set_xlabel(feat)
        plt.tight_layout()
        fig.savefig(carpeta / f"hist_{feat}.png", dpi=120)
        plt.close(fig)

    # 2. Heatmap de correlaciones entre features
    fig, ax = plt.subplots(figsize=(8, 6))
    corr = df[features].corr()
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", ax=ax)
    ax.set_title("Correlaciones entre features")
    plt.tight_layout()
    fig.savefig(carpeta / "heatmap_correlaciones.png", dpi=120)
    plt.show()

    # 3. Barplot de correlación de cada feature con el target
    corr_target = df[features + [target]].corr()[target].drop(target)
    fig, ax = plt.subplots(figsize=(7, 4))
    corr_target.sort_values().plot(kind="barh", ax=ax)
    ax.set_title(f"Correlación con {target}")
    ax.axvline(0, color="gray", linestyle="--")
    plt.tight_layout()
    fig.savefig(carpeta / f"corr_{target}.png", dpi=120)
    plt.show()

# Prueba con Titanic
titanic = sns.load_dataset("titanic")
titanic = titanic.dropna(subset=["embarked"])
titanic["age"] = titanic["age"].fillna(titanic["age"].median())

features = ["age", "fare", "pclass", "sibsp", "parch"]
eda_rapido(titanic, features, "survived")

```
