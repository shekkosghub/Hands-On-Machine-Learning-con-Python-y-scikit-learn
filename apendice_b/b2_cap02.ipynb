```python
# ---------------------------------------------------------------
# CAPÍTULO 2/numpy — Ejercicio 1
# ---------------------------------------------------------------

import numpy as np
precios = np.array([19.99, 34.50, 12.00, 89.99, 5.49, 47.00])
print(f"Media:   {np.mean(precios):.2f}") 
print(f"Mínimo:  {np.min(precios):.2f}") 
print(f"Máximo:  {np.max(precios):.2f}") 
print(f"Rango:   {np.max(precios) - np.min(precios):.2f}") 
# Descuento 15% — operación vectorizada sobre todo el array
precios_desc = precios * 0.85 
print("\nPrecios con 15% de descuento:") 
print(precios_desc.round(2))
```

```python
# ---------------------------------------------------------------
# CAPÍTULO 2/numpy — Ejercicio 2
# ---------------------------------------------------------------

import numpy as np
rng = np.random.default_rng(42)
X = rng.standard_normal((5, 4))       # 5 ejemplos, 4 features
# a) Tercera columna (índice 2)
col3 = X[:, 2]
print("Tercera columna:", col3.round(3))
# b) Últimas 2 filas
ultimas_2 = X[-2:, :]
print("Últimas 2 filas:", ultimas_2.shape)
# c) Normalización manual (media=0, std=1) por columna
media_col  = X.mean(axis=0)
std_col    = X.std(axis=0)
X_norm     = (X - media_col) / std_col
# d) Verificar que la media de cada columna sea ≈ 0
print("Media por columna (debería ser ≈ 0):")
print(X_norm.mean(axis=0).round(10))
```

```python
# ---------------------------------------------------------------
# CAPÍTULO 2/numpy — Ejercicio 3
# ---------------------------------------------------------------

# Celda 1 — Cargar y explorar
import numpy as np
from sklearn.datasets import fetch_california_housing
california = fetch_california_housing()
X = california.data
print(f"Shape: {X.shape}")
print("Media por feature:  ", X.mean(axis=0).round(3))
print("Std por feature:    ", X.std(axis=0).round(3))
# Slicing: primeras 100 filas, primeras 3 columnas
X_sub = X[:100, :3]
print(f"Subconjunto shape:  {X_sub.shape}")
```

```python
# Celda 2 — Medir np.sum con %%timeit
%%timeit
np.sum(X)
```

```python
# Celda 3 — Medir bucle equivalente con %%timeit
%%timeit
total = 0
for fila in X:
    for valor in fila:
        total += valor
# NumPy suele ser 50–200× más rápido que este bucle
```
